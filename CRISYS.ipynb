```python
# Function to automatically condense the original prompt based on specified criteria
def condense_prompt(input_data):
    # Initialize an empty list for the condensed prompt parts
    condensed_parts = []

    # Extract and condense input based on the given categories
    if "Sketch Type" in input_data:
        condensed_parts.append("ultra-Natural black and white pencil face portrait.")
    if "Face Type" in input_data:
        condensed_parts.append(input_data["Face Type"])
    if "Eye Type" in input_data:
        condensed_parts.append(input_data["Eye Type"])
    if "Nose Shape" in input_data:
        condensed_parts.append(input_data["Nose Shape"])
    if "Body Color" in input_data:
        condensed_parts.append(input_data["Body Color"])
    if "Ethnicity" in input_data:
        condensed_parts.append(input_data["Ethnicity"])
    if "Hair Style" in input_data:
        condensed_parts.append(input_data["Hair Style"])
    if "Facial Expression" in input_data:
        condensed_parts.append(input_data["Facial Expression"])
    if "Age Group" in input_data:
        condensed_parts.append(input_data["Age Group"])
    if "Miscellaneous Details" in input_data:
        condensed_parts.append(input_data["Miscellaneous Details"])

    # Join the condensed parts into a single prompt
    return " ".join(condensed_parts)

# Function to handle image generation from user input
def generate_image_from_input(user_input):
    # Automatically condense the user input
    condensed_prompt = condense_prompt(user_input)

    # Generate the image with the condensed prompt
    image = pipe(condensed_prompt).images[0]

    # Display the generated image
    import matplotlib.pyplot as plt
    plt.imshow(image)
    plt.axis("off")
    plt.show()

    # Save the final image if needed
    image.save("final_combined_sketch.png")

# Example of user input (can be replaced with actual user input)
user_input = {
    "Sketch Type": "ultra-Natural,black and white pencil face portrait",
    "Face Type": "Round face with full cheeks.",
    "Eye Type": "Large almond-shaped eyes.",
    "Nose Shape": "Broad flat nose.",
    "Body Color": "Deep, rich ebony tones.",
    "Ethnicity": "African with pronounced features.",
    "Hair Style": "Short, tightly curled Afro.",
    "Facial Expression": "Startled surprise: mouth slightly open, raised eyebrows.",
    "Age Group": "Teenager with youthful skin.",
    "Miscellaneous Details": "Wearing thick, black-rimmed glasses slightly askew."
}

# Automatically generate the image using user input
generate_image_from_input(user_input)
```

```python


def process_prompt():
    last_prompt = ""  # To store the last prompt entered
    last_image = None  # To store the last generated image

    while True:
        # Take input for the prompt, allowing the user to modify the last one
        if last_prompt:
            prompt = input(f"Modify your previous prompt '{last_prompt}' or enter a new one: ")
        else:
            prompt = input("Enter your prompt: ")

        # Append the new prompt to the last prompt if modifying
        if prompt:
            final_prompt = last_prompt + " " + prompt
        else:
            final_prompt = last_prompt

        # Process the prompt here (you can replace this with actual logic)
        print(f"Processing your prompt: {final_prompt}")
        image = pipe(final_prompt).images[0]  # Assuming this generates an image

        # Ask if the prompt is okay
        while True:
            retry = input("Is the prompt okay? (y/n): ").lower()

            if retry == 'y':
                print("Prompt accepted!")
                last_prompt = final_prompt  # Store the final prompt
                last_image = image  # Store the last image for reference
                plt.imshow(image)
                plt.axis("off")
                plt.show()
                break
            elif retry == 'n':
                print("Retrying the prompt...")
                break
            else:
                print("Invalid input! Please enter 'y' for yes or 'n' for no.")

        # If the prompt is okay, ask whether to continue or exit
        if retry == 'y':
            continue_input = input("Do you want to input another prompt? (y/n): ").lower()

            if continue_input == 'n':
                print("Exiting the prompt input process.")
                break
            elif continue_input == 'y':
                print("You can modify the previous prompt in the next step.")
                continue
            else:
                print("Invalid input! Please enter 'y' for yes or 'n' for no.")
        else:
            # Retry input, keeping the previous prompt
            continue

if __name__ == "__main__":
    process_prompt()
```
